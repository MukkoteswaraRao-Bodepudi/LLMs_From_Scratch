```python
#import the library
import tiktoken
```

```python
## initialize the tokenizer from tiktoken
tokenizer = tiktoken.get_encoding("gpt2")
```

```python
### Data
text = "The quick brown fox jumps over the lazy dog."

##conver the text into token Id's

integers = tokenizer.encode(text)
print(integers)
```

```python
## let's decode the each token at a time using loop

for token in integers:
    print(token,":",tokenizer.decode_single_token_bytes(token))
```

```python
## Let's Test Uncommon Word

text = "unbelievable"
integers = tokenizer.encode(text)
print(integers)

for token in integers:
    print(token,":",tokenizer.decode_single_token_bytes(token))

## lets decode as single text
text = tokenizer.decode(integers)
print(text)
```

#### **Compare how BPE tokenizes words**

```python
##Step 1 — Create a list of words
words = [
    "play",
    "playing",
    "played",
    "player",
    "playground",
    "unplayable"
]
```

```python
##Step 2 — Tokenize each word
for word in words:

    integer = tokenizer.encode(word)

    print("Words:",word)
    print("Token ID's:",[integer])

    ## decode
    for token in integer:
        print(token,":",tokenizer.decode_single_token_bytes(token))
```

```python
#count number of tokens for each word
for word in words:

    integers = tokenizer.encode(word)

    print(
        f"{word:12} → {len(integers)} tokens"
    )
```

#### ***Task 3: the same experiment with sentences,***

```python
### List of sentences
sentences = [
    "I play football.",
    "I am playing football.",
    "I played football yesterday.",
    "The player is playing.",
    "The playground is large.",
    "This game is unbelievable."
]
```

```python
##Step 2 — Tokenize each word
for sentence in sentences:

    integer = tokenizer.encode(sentence)

    print("Words:",sentence)
    print("Token ID's:",[integer])

    ## decode
    for token in integer:
        print(token,":",tokenizer.decode_single_token_bytes(token))
```

```python
#count number of tokens for each word
for sentence in sentences:

    integers = tokenizer.encode(sentence)

    print(
        f"{sentence:30} → {len(integers)} tokens"
    )
```

```python
### decode and verify
for sentence in sentences:

    integers = tokenizer.encode(sentence)
    decode = tokenizer.decode(integers)

    print("Original:",sentence)
    print("Decoded:",decode)
    print("Match: ",decode==sentence)
```
