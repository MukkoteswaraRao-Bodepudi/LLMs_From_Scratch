#### **Token Embeddings**
##### **Token Embeddings are converts each token id into dense numerical vector representation that are neural network can learn from.**

```python
import torch

#small vocabulary
vocab = {
    "I": 0,
    "love": 1,
    "LLMs": 2,
    "learning": 3,
    "Python": 4
}

## create token ids
input_ids = torch.tensor([0,1,2,3])
```

```python
## create a embedding layer using vocab size and size of each vector is 4
import torch.nn as nn

vocab_size = len(vocab)
embedding_dim = 4

#embedding layer
torch.manual_seed(123)
embedding_layer = nn.Embedding(vocab_size,embedding_dim)
print(embedding_layer)
```

```python
## weights
embedding_layer.weight
```

```python
# Get embeddings for input_ids
input_ids
```

```python
# Print the embeddings for input_ids
embedding_layer(input_ids)
```

```python
# Print the shape of the embeddings
embedding = embedding_layer(torch.tensor(list(vocab.values())))
embedding.shape
```

```python
# Check whether the embedding weights are trainable
embedding_layer.weight.requires_grad
```
