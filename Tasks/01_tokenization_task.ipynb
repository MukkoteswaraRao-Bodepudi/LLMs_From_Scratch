```python
## Load the Data from the File. 

with open(path,'r',encoding = "utf-8") as f:
    raw_data = f.read()


print("Number Of Characters:",len(raw_data))
print(raw_data[:99])
```

### ***Break down the larger data into tokens by encountering the space in the data***

```python
import re
test_data = "Hello World, This is an Example for Test."
result = re.split(r'(\s)',test_data)
print(result)
```

```python
## Lets splits the punctuation also like ,.! and more
result = re.split(r'([,.:;!?_"()\']|--|\s)',test_data)
print(result)
```

```python
## remove the white spaces to reduce the memory and computing requirements. 
preprocessed = [item for item in result if item.strip()]
preprocessed
```

```python
## Let's apply same logic to original data to breakdown into tokens. 
preprocessed = re.split(r'([,.:;!?_"()\']|--|\s)',raw_data)
preprocessed = [item for item in preprocessed if item.strip()]
print(len(preprocessed))
```

```python
## unique tokens in ascending order by vocabulary Alphabets
all_tokens = sorted(set(preprocessed))
print("Unique Tokens:",len(all_tokens))
```

```python
### Let's handle the issue like unknown word and end of texts
all_tokens.extend(['<|unk|>','<|endoftext|>'])
print(len(all_tokens))
```

```python
###create a mapping like token is map to the token id 
vocab = {token:integer for integer,token in enumerate(all_tokens)}

for i,item in enumerate(vocab.items()):
    print(item)

    if i>=10:
        break
```

```python
for i,item in enumerate(list(vocab.items())[-5:]):
    print(item)
```

```python
### Create a class 
class Tokenization:
    def __init__(self,vocab):
        self.str_to_int = vocab
        self.int_to_str = {integer:token for token,integer in vocab.items()}

    ### Encoder is helps to convert text into token ids
    def encoder(self,text):
        preprocessed = re.split(r'([,.:;!?_"()\']|--|\s|<\|endoftext\|>)',text)
        preprocessed = [item for item in preprocessed if item.strip()]
        preprocessed = [item if item in self.str_to_int else "<|unk|>" for item in preprocessed]
        ids = [self.str_to_int[s] for s in preprocessed]
        return ids

    ### Decoder it helps to convert the ids into text
    def decoder(self,ids):
        text = " ".join([self.int_to_str[id] for id in ids])

        ## remove the white spaces before special punctuations
        cleaned_text = re.sub(r'\s+([,.:;!?_"()\'])',r"\1",text)

        return cleaned_text
```

```python
tokenizer = Tokenization(vocab)

text = "Large language models learn patterns from text.ss"

ids = tokenizer.encoder(text)
print(ids)
```

```python
text = tokenizer.decoder(ids)
print(ids)
print(text)
```
