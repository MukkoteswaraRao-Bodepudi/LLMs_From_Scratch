### ***Token Embeddings***

#### **Token Embeddings means converts each token id into dense numerical vector that the neural network can learn from.**

##### ***Word2vec-google-news-300***
##### Example Pre-trained Token Embedding Model.
##### Dataset contains roughly 300 billion words
##### 3 Million Unique Words
##### 300 Dimension Vector

```python
import gensim.downloader as api

model = api.load("word2vec-google-news-300")
```

```python
vocab_vector = model
## Model is dataset like input as Word and output as dense Numberical Vector.
```

```python
### Let's test the Number Of Dimensional of each vector
print(len(vocab_vector['King']))
```

```python
print(len(vocab_vector))
```

#### **Similar Words**

```python
### get similar words for King+Woman-Man = ?

print(vocab_vector.most_similar(positive=['King','Woman'],negative=['Man'],topn=10))
```

```python
### Let's similarity between Words
print(vocab_vector.similarity("Cat","Dog"))
print(vocab_vector.similarity("Milk","Tea"))
print(vocab_vector.similarity("boy","girl"))
print(vocab_vector.similarity("keyboard","desktop"))
```

```python
## Lets's calculate the magnitude of two words

import numpy as np 

word1 = "woman"
word2 = "man"

word3 = "dog"
word4 = "cat"

word5 = "book"
word6 = "pen"

###Calculate the Vector Difference
vector_difference1 = model[word1]-model[word2]
vector_difference2 = model[word3]-model[word4]
vector_difference3 = model[word5]-model[word6]

### Magnitude
magnitude1 = np.linalg.norm(vector_difference1)
magnitude2 = np.linalg.norm(vector_difference2)
magnitude3 = np.linalg.norm(vector_difference3)

print(f"Magnitude of {word1} and {word2} is:",magnitude1)
print(f"Magnitude of {word3} and {word4} is:",magnitude2)
print(f"Magnitude of {word5} and {word6} is:",magnitude3)
```

```python
### Let's build embedding weights matrix with an Example. Suppose Input Tokens are 2,3,5, and 1.
import torch
input_ids = torch.tensor([2,3,5,1]) #Every Token ID associated with a word

```

```python
## for Simple Embedding Weights matrix, let's take the size of vocab is 6 and vector dim is 3
## Gpt-2 was trained on 50,257 tokens as vocab size and vector dim is 768. 
```

```python
vocab_size = 6
output_dim = 3 #size of each token
```

```python
#Using the Pytorch we can initalize the Embedding Layer by passing the vocab size and output dim ,
#it will create a lookup table like embedding weight matrix with vocab size as row and output dim as column.
torch.manual_seed(123) #manual seed is fixed will not change for every execution.
embedding_layer = torch.nn.Embedding(vocab_size,output_dim)
print(embedding_layer)
```

```python
### let's see the weights
embedding_layer.weight
```

```python
### Weights of Embedding layer consists of random value's, 
# these values will optimize during the LLM Training
```

```python
##let's get the vector using the token id is 3
print(embedding_layer(torch.tensor([3])))
```

```python
print(embedding_layer(input_ids))
```
