#### ***BPE(Byte Pair Encoding)***

##### **Byte Pair Encoding is a subword tokenization algorithm that starts with small units and repeatedly merges frequently occurring adjacent pairs to build a vocabulary of reusable tokens.**

### ***TikToken***
#### **tiktoken is a tokenizer library developed by OpenAI. It converts text into tokens, and then converts those tokens into integer IDs that an LLM can process.**

```python
import tiktoken
```

```python
tokenizer = tiktoken.get_encoding("gpt2")
```

#### **Converts token into token Id's**

```python
text = """Hello, do you like tea? <|endoftext|>
    In the sunlit terraces of the someUnknownplace."""


integers = tokenizer.encode(text,allowed_special={"<|endoftext|>"})
print(integers)
```

### **Decoder: Converts the Token Id's into Tokens**

```python
text = tokenizer.decode(integers)
text
```

#### ***Let's test how the gpt-2/BPE tokenizer dealing with the unknown words***


```python
integers = tokenizer.encode("Arkdje tr")
print(integers)

text = tokenizer.decode(integers)
print(text)
```
