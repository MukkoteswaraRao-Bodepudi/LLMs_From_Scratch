#### **1.Tokenization**
##### ***The Process of Break down the raw data into small chunks called Tokens, and assign Id's to them called Token Id's.***

```python
### Load the Data From the External File

with open("../Data/the-verdict.txt","r",encoding = "utf-8") as f:
    raw_text = f.read()

print("Number of Characters:",len(raw_text))
print("Last 100 Characters:",raw_text[-100:])
```

### **1.Word Level Tokenization**
#### **It's splits the raw data into words.**

```python
import re

#step-1: splits the data into words when space is encountered.
test_data = "Wow! Natural-language processing (NLP) isn't easy, is it?"

words = re.split(r'(\s)',test_data)
print(words)
```

```python
## step-2: Based on step-1 punctuations and some extra characters are with words, let's Encounter. 
words = re.split(r'([,.:;"_/?!()\']|--|\s)',test_data)
print(words)
```

```python
### step-3: Based on Step 2 let's remove the white spaces, those are not useful to us.
words = [word for word in words if word.strip()]
print(words)
```

```python
## Let's apply to the original data

preprocessed_data = re.split(r'([,.:;"_/?!()\']|--|\s)',raw_text)
preprocessed_data = [data for data in preprocessed_data if data.strip()]
print(len(preprocessed_data))
```

```python
## get the unique words and arrange them in alphabetical order

all_words = sorted(set(preprocessed_data))

## assign a token id's to each token
vocab = {token:integer for integer,token in enumerate(all_words)}

for i, token in enumerate(vocab.items()):
    print(token)

    if i>=50:
        break
```

```python
### Let's build simple Tokenizer

class SimpleTokenizerv1:

    def __init__(self,vocab):
        self.str_to_int = vocab
        self.int_to_str = {value:key for key,value in vocab.items()}

    ## Encoder convert the text into token id's 
    def encode(self,txt):
        preprocessed_data = re.split(r'([,.:;"_/?!()\']|--|\s)',txt)
        preprocessed_data = [data for data in preprocessed_data if data.strip()]
        ids = [self.str_to_int[s] for s in preprocessed_data]
        return ids

    #decoder
    def decode(self,ids):
        text = " ".join([self.int_to_str[id] for id in ids])
        ##clean the text because it add's the space for special characters also
        clean_text = re.sub(r'\s+([,.:;"_/?!()\'])',r'\1',text)

        return clean_text
        
```

```python
tokenizer = SimpleTokenizerv1(vocab)

sample_text = "Hang it, Rickham, with that face watching me I couldn't do another stroke."

token_ids = tokenizer.encode(sample_text)
print(token_ids)
```

```python
clean_txt = tokenizer.decode(token_ids)
print(clean_txt)
```

```python
### Let's build simple Tokenizer

class SimpleTokenizerv2:

    def __init__(self,vocab):
        self.str_to_int = vocab
        self.int_to_str = {value:key for key,value in vocab.items()}

    ## Encoder convert the text into token id's 
    def encode(self,txt):
        preprocessed_data = re.split(r'([,.:;"_/?!()\']|--|\s|<\|endoftext\|>)',txt)
        preprocessed_data = [data for data in preprocessed_data if data.strip()]
        preprocessed_data = [data if data in self.str_to_int else '<|unk|>' for data in preprocessed_data]
        ids = [self.str_to_int[s] for s in preprocessed_data]
        return ids

    #decoder
    def decode(self,ids):
        text = " ".join([self.int_to_str[id] for id in ids])
        ##clean the text because it add's the space for special characters also
        clean_text = re.sub(r'\s+([,.:;"_/?!()\'])',r'\1',text)

        return clean_text
        
```

#### ***Create a Input-Target Pair***

#### **Input-Target Pairs in LLMs teach the model How to predict the next token in a sequence of text.**

```python
### create a dataset to organize the data into input and target--> Training Examples
import torch
from torch.utils.data import DataLoader,Dataset
class GptDatasetV1(Dataset):

    def __init__(self,txt,tokenizer,max_length,stride):
        self.input_ids = []
        self.target_ids = []

        token_ids = tokenizer.encode(txt)

        for i in range(0,len(token_ids),stride):
            input_chunk = token_ids[i:i+max_length]
            target_chunk = token_ids[i+1:i+max_length+1]
            self.input_ids.append(torch.tensor(input_chunk))
            self.target_ids.append(torch.tensor(target_chunk))

    def __len__(self):
        return len(self.input_ids)

    def __getitem__(self,idx):
        return self.input_ids[idx],self.target_ids[idx]
```

```python
### How the input and target are stored in example==> Example = input and target
```

```python
### Dataloader used to retrieve the examples from the Dataset.

def create_dataloader(txt,batch_size = 8,max_length=4,stride=1,shuffle=True,drop_last=True,num_workers = 0):

    #Initailize the Tokenizer
    tokenizer = tiktoken.get_encoding("gpt2")

    dataset = GptDatasetV1(txt,tokenizer,max_length = max_length,stride = stride)

    dataloader = DataLoader(dataset,
                batch_size=batch_size,shuffle = shuffle,drop_last=drop_last,num_workers=0)

    return dataloader
```

```python
dataloader = create_dataloader(raw_text,batch_size = 8, max_length = 6, stride = 3,shuffle = False)

batch_iter = iter(dataloader)
first_batch = next(batch_iter)

input,target = first_batch

print("Input:",input)
print('Target:',target)
```

```python
## lets decode the first example from first batch
first_input = input[3].tolist()
first_target = target[3].tolist()

input_text = bpe_tokenizer.decode(first_input)
target_text = bpe_tokenizer.decode(first_target)

print("Input:",input_text)
print("Target:",target_text)
```

#### ***Token Embedding***
##### ***The process of coverting token id's into the Embedding Vectors.***

```python
### let's build a embedding layer based on the vocab size and dimensional
import torch.nn as nn

vocab_size = 50257
out_dim = 256

embedding_layer = nn.Embedding(vocab_size,out_dim)

print(embedding_layer)
```

```python
###lets build token embedding using embedding layer using sample input
token_embedding = embedding_layer(input)
print(token_embedding.shape)
```

#### ***Position Embedding***
#### ***It tells the LLM what the exact position and sequence of the text.***

```python
###using absoulte position Embedding means it gives the unique position the each token in input sample

### we have 6 columns in each row based on this let's build the position Embedding 6x256

pos_embedding_layer = nn.Embedding(6,256)

print(pos_embedding_layer)
```

```python
pos_embedding = pos_embedding_layer(torch.arange(6)) #0,1,2,3,4,5
print(pos_embedding.shape)
```

#### ***Input Embedding before training the LLM, it tells the model what exact position of token in a sequence text using token embedding and position embedding***

```python
input_embedding = token_embedding+pos_embedding
print(input_embedding)
```
