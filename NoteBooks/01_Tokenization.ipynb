```python
### valid the file path and load the data from the file


#Check the path is existed or not
import os
path = "../Data/the-verdict.txt"
if os.path.exists(path):
    print("Path is Existed")
else:
    print("Path is not Exist.")

with open(path,"r",encoding="utf-8") as f:
    raw_text = f.read()

print("Number OF Characters:",len(raw_text))
print(raw_text[:300])
```

```python
### Let's split's the data based when white space is encountered using re(regular expression).
import re

test_data = "Hello, World?. This - is an Example-- to split the Data."

result = re.split(r'(\s)',test_data)
print(result)
```

```python
## And also split the data further into commas and . 
result = re.split(r'([,.:;-?_!"()\']|--|\s)',test_data)
print(result)
```

```python
### whitespaces are also treatings as token, let's remove
result = [item for item in result if item.strip()] 
#Removing whitespace's reduce the memory and computing requirements.
print(result)
```

```python
### let's apply the regular expression to the original data
preprocessed = re.split(r'([,.:;-?_!"()\']|--|\s)',raw_text)
preprocessed = [item for item in preprocessed if item.strip()] 
print(preprocessed[:30])
```

```python
### Let's Count Number of Tokens
print(len(preprocessed))
```

#### **Create Token ID's**

```python
### normal text contains the duplicate right using we can handle and sort the order
all_words = sorted(set(preprocessed))


##number of words
vocab_size = len(all_words)

print("vocabulary size:",vocab_size)
```

```python
### Create id's and assign to the token's and stored it in a dictionary

vocab = {token:integer for integer,token in enumerate(all_words)}

### Here we succesfully build or mapped the token to token id's for encoder setup.
```

```python
## let's print
for i,item in enumerate(vocab.items()):
    print(item)
    if i>=50:
        break
```

```python
## For decoder we have to get the token based on the token id, we have mapped it properly for decoder usage
#let's do that.set
decoder_vocab  = {value:key for key,value in vocab.items()}

for i,item in enumerate(decoder_vocab.items()):
    print(item)
    if i>=50:
        break
```

```python
class SimpleTokenizerV1:
    def __init__(self,vocab):
        self.str_to_int = vocab
        self.int_to_str = {value:key for key,value in vocab.items()}

    def encode(self,text):

        preprocessed = re.split(r'([,.:;-?_!"()\']|--|\s)',text)

        preprocessed = [item for item in preprocessed if item.strip()]

        ids = [self.str_to_int[s] for s in preprocessed]

        return ids

    def decode(self,ids):

        text = " ".join([self.int_to_str[i] for i in ids])

        ## Remove the spaces before the specific punctuations. 
        cleaned_text = re.sub(r"\s+([,.:;\-?_!\"()'])", r"\1", text)

        return cleaned_text
```

```python
tokenizer = SimpleTokenizerV1(vocab)

text = """It's the last he painted, you know,
       Mrs. Gisburn said with pardonable pride."""
    
ids = tokenizer.encode(text)
print(ids)
```

```python
text = tokenizer.decode(ids)
print(text)
```

#### **Adding Special Context Tokens**

```python
### we will implement a new class which will handle the unknown tokens.
### and also we have handle like when we working multiple source text,
#  we have endoftext token between the unrelated texts like to identify marker,and other. 
all_tokens = sorted(set(preprocessed))
all_tokens.extend(["<|endoftext|>", "<|unk|>"])



vocab = {token: integer for integer, token in enumerate(all_tokens)}
```

```python
## lets replace the unknown word by <|unk|> token
```

```python
class SimpleTokenizerV2:
    def __init__(self,vocab):
        self.str_to_int = vocab
        self.int_to_str = {value:key for key,value in vocab.items()}

    def encode(self,text):

        preprocessed =  re.split(r'([,.:;?_!"()\']|--|\s|<\|endoftext\|>)',text)

        preprocessed = [item for item in preprocessed if item.strip()]

        preprocessed = [item if item in self.str_to_int else "<|unk|>" for item in preprocessed]

        ids = [self.str_to_int[s] for s in preprocessed ]

        return ids

    def decode(self,ids):

        text = " ".join([self.int_to_str[i] for i in ids])

        ## Remove the spaces before the specific punctuations. 
        cleaned_text = re.sub(r"\s+([,.:;\-?_!\"()'])", r"\1", text)

        return cleaned_text
```

```python
text1 = "Hello, do you like tea?"
text2 = "In the sunlit terraces of the palaces."
text = " <|endoftext|> ".join((text1,text2))
print(text)
tokenizer = SimpleTokenizerV2(vocab)
ids = tokenizer.encode(text)
print(ids)
```

```python
text = tokenizer.decode(ids)
print(text)
```
